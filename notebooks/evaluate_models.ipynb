# SignBridge AI — Model Evaluation

Load the trained models in `models/asl_models.pkl` and compare the three
classifiers (Support Vector Machine, Random Forest, Logistic Regression)
against the held-out test set.

```python
import sys, os, joblib
sys.path.insert(0, os.getcwd())

from ai._paths import ASL_MODELS_PATH, PROCESSED_DATA_PATH

models = joblib.load(ASL_MODELS_PATH)
data = joblib.load(PROCESSED_DATA_PATH)

X_test = data["X_test"]
y_test = data["y_test"]
print("Classes:", list(models["label_encoder"].classes_))
print("Test size:", X_test.shape)
```

```python
from sklearn.metrics import accuracy_score

for name in ("svm", "random_forest", "logistic_regression"):
    clf = models[name]
    pred = clf.predict(X_test)
    print(f"{name:20s} -> {accuracy_score(y_test, pred) * 100:.2f}%")
```
